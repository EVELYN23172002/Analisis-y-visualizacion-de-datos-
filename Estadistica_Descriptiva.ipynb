#Práctica 1.3 Estadística Descriptiva
Referencia: https://rpubs.com/CristianPerdomo/estadisticadescriptiva

```r
install.packages("readxl")
library(readxl)
```

```r
datos <- read_excel("data.xlsx")
head(datos)
```

#1.1 Medidas de tendencia central

1.1.1 Media

```r
mean(datos$Estatura)
```

```r
apply(as.matrix(datos[,2:4]), 2,mean)
```

```r
lapply(datos[,2:4], mean)
```

```r
unlist(lapply(datos[,2:4], mean)) # unlist -> para no verlo como lista
```

1.1.2 Mediana



```r
median(datos$Estatura)
```

```r
apply(as.matrix(datos[,2:4]),2, median)
```

```r
unlist(lapply(datos[,2:4], median)) # unlist -> para no verlo como lista
```

#1.2 Medidas de dispersión

1.2.1 Varianza y desviación estándar

```r
var(datos$Estatura)
```

```r
sd(datos$peso)
```

```r
apply(as.matrix(datos[,2:4]), 2, sd)
```

```r
unlist(lapply(datos[,2:4], sd)) # unlist -> para no verlo como lista
```

#1.3 Medidas de localización

1.3.1 Cuantiles

```r
quantile(datos$Estatura)
```

```r
quantile(datos$Estatura, probs = 0.25) #-> Si solo deceamos uno en específico.
```

```r
quantile(datos$Estatura, probs = c(0.25, 0.75)) #-> dos o mas utilizamos c(valor1, valor2,...)
```

1.3.2 Percentiles

```r
quantile(datos$Estatura, probs = 0.20)
```

```r
quantile(datos$Estatura, probs = c(0.10, 0.15 ,0.20))
```

#1.4 Adicionales

```r
summary(datos)
```

```r
datos$Helado = as.factor(datos$Helado)
summary(datos)
```

#CONCLUSIONES DE LA PRÁCTICA

El documento presenta el uso de comandos básicos de estadística descriptiva en R, cuyo objetivo principal es analizar, resumir e interpretar un conjunto de datos a partir de medidas estadísticas fundamentales. A lo largo del ejercicio se trabajan datos numéricos como estatura y peso, así como una variable categórica, lo que permite comprender distintos tipos de análisis.

En primer lugar, se utilizan los comandos install.packages("readxl") y library(readxl) para instalar y cargar la librería necesaria que permite importar datos desde archivos Excel. Posteriormente, con read_excel("data.xlsx"), se cargan los datos al entorno de R, y con head(datos) se visualizan las primeras filas para verificar que la información se haya importado correctamente.

Después, se emplean comandos como mean(), median(), var() y sd() para calcular medidas de tendencia central y de dispersión.

mean() permite conocer el promedio,

median() indica el valor central,

var() muestra la variabilidad de los datos,

sd() calcula la desviación estándar, es decir, qué tanto se alejan los valores del promedio.

Para analizar varias columnas al mismo tiempo, se utilizan funciones como apply() y lapply(), las cuales automatizan los cálculos estadísticos sobre diferentes variables numéricas. El uso de unlist() es importante porque convierte los resultados en un formato más sencillo y legible, evitando que se muestren como listas.

Además, el uso del comando quantile() permite identificar la distribución de los datos, obteniendo percentiles específicos como el 25%, 50% o 75%. Esto es clave para entender cómo se concentran los valores y detectar posibles valores atípicos.

El comando summary(datos) ofrece un resumen estadístico general, combinando mínimo, máximo, cuartiles, media y, en el caso de variables categóricas, la frecuencia de cada categoría. Finalmente, al convertir la variable Helado en un factor con as.factor(), se logra que R reconozca correctamente los datos cualitativos, mejorando su interpretación en el resumen estadístico.
